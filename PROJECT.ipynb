```python
# Step 1: Separate truly continuous variables from categorical/discrete ones
continuous_cols = [
    'age_at_enrollment',
    'curricular_units_1st_sem_(credited)',
    'curricular_units_1st_sem_(enrolled)',
    'curricular_units_1st_sem_(evaluations)',
    'curricular_units_1st_sem_(approved)',
    'curricular_units_1st_sem_(grade)',
    'curricular_units_1st_sem_(without_evaluations)',
    'curricular_units_2nd_sem_(credited)',
    'curricular_units_2nd_sem_(enrolled)',
    'curricular_units_2nd_sem_(evaluations)',
    'curricular_units_2nd_sem_(approved)',
    'curricular_units_2nd_sem_(grade)',
    'curricular_units_2nd_sem_(without_evaluations)',
    'unemployment_rate',
    'inflation_rate',
    'gdp'
]

categorical_cols = [
    'marital_status',
    'application_mode',
    'application_order',
    'course',
    'daytime/evening_attendance',
    'previous_qualification',
    'nacionality',
    "mother's_qualification",
    "father's_qualification",
    "mother's_occupation",
    "father's_occupation",
    'displaced',
    'educational_special_needs',
    'debtor',
    'tuition_fees_up_to_date',
    'gender',
    'scholarship_holder',
    'international'
]

print(f"Continuous variables: {len(continuous_cols)}")
print(f"Categorical variables: {len(categorical_cols)}")
```

```python
# Function to detect outliers using IQR method
def detect_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers_mask = (df[column] < lower_bound) | (df[column] > upper_bound)
    n_outliers = outliers_mask.sum()
    outlier_percentage = (n_outliers / len(df)) * 100
    
    return {
        'n_outliers': n_outliers,
        'percentage': outlier_percentage,
        'lower_bound': lower_bound,
        'upper_bound': upper_bound,
        'min_value': df[column].min(),
        'max_value': df[column].max()
    }

# Check outliers only in continuous variables
outlier_summary = []
for col in continuous_cols:
    result = detect_outliers(X, col)
    outlier_summary.append({
        'feature': col,
        **result
    })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df = outlier_df[outlier_df['n_outliers'] > 0].sort_values('percentage', ascending=False)

print("Outliers detected in continuous variables:\n")
print(outlier_df.to_string(index=False))
```

```python
# Visualize outliers in continuous variables with most outliers
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

top_outlier_cols = outlier_df.head(9)['feature'].tolist() if len(outlier_df) >= 9 else continuous_cols[:9]

for idx, col in enumerate(top_outlier_cols):
    if idx < len(axes):
        sns.boxplot(data=X, y=col, ax=axes[idx], color='lightblue')
        axes[idx].set_title(f'{col}\n({outlier_df[outlier_df["feature"]==col]["n_outliers"].values[0] if col in outlier_df["feature"].values else 0} outliers)', 
                           fontsize=10)
        axes[idx].tick_params(labelsize=8)

plt.tight_layout()
plt.show()
```

```python
# Run this diagnostic first
print("="*70)
print("CATEGORICAL VARIABLE DIAGNOSTIC FOR STUDENT RETENTION PROJECT")
print("="*70)

issues_found = {
    'rare_categories': [],
    'highly_imbalanced': [],
    'low_variance': []
}

for col in categorical_cols:
    value_counts = df[col].value_counts()
    value_props = df[col].value_counts(normalize=True)
    
    print(f"\n{col}:")
    print(f"  Unique values: {df[col].nunique()}")
    
    # Check 1: Rare categories (< 1%)
    rare = value_props[value_props < 0.01]
    if len(rare) > 0:
        print(f"  ⚠️ {len(rare)} rare categories (< 1%): {rare.index.tolist()}")
        issues_found['rare_categories'].append(col)
    
    # Check 2: Highly imbalanced (one category > 95%)
    if value_props.max() > 0.95:
        dominant = value_props.idxmax()
        print(f"  ⚠️ HIGHLY IMBALANCED: {value_props.max()*100:.1f}% are '{dominant}'")
        issues_found['highly_imbalanced'].append(col)
    
    # Check 3: Low variance (one category > 98%)
    if value_props.max() > 0.98:
        print(f"  ⚠️ LOW VARIANCE: Nearly constant feature")
        issues_found['low_variance'].append(col)
    
    # Show distribution
    print(f"  Top 3 categories:")
    for i, (value, prop) in enumerate(value_props.head(3).items()):
        print(f"    {value}: {prop*100:.1f}%")

print("\n" + "="*70)
print("SUMMARY")
print("="*70)
print(f"Features with rare categories: {len(issues_found['rare_categories'])}")
print(f"Features highly imbalanced: {len(issues_found['highly_imbalanced'])}")
print(f"Features with low variance: {len(issues_found['low_variance'])}")
```

```python
# First, let's identify which feature has low variance
print("Identifying low-variance features (>98% in one category):\n")

low_variance_features = []
for col in categorical_cols:
    max_prop = df[col].value_counts(normalize=True).max()
    if max_prop > 0.98:
        low_variance_features.append(col)
        print(f"❌ {col}: {max_prop*100:.2f}% in one category")

print(f"\nFeatures to drop: {low_variance_features}")

# Drop the low-variance feature(s)
X_processed = X.drop(columns=low_variance_features)
categorical_cols_updated = [c for c in categorical_cols if c not in low_variance_features]

print(f"\n✓ Dropped {len(low_variance_features)} low-variance feature(s)")
print(f"Features remaining: {X_processed.shape[1]} (was {X.shape[1]})")
```

```python
# Only group rare categories in features with >10 unique values
print("\n" + "="*70)
print("Handling rare categories in high-cardinality features:")
print("="*70)

def group_rare_categories(df, col, threshold=0.01, min_cardinality=10):
    """
    Group rare categories into 'Other' category
    Only for features with >min_cardinality unique values
    """
    if df[col].nunique() <= min_cardinality:
        print(f"✓ {col}: {df[col].nunique()} categories - keeping all (manageable)")
        return df
    
    value_props = df[col].value_counts(normalize=True)
    rare_categories = value_props[value_props < threshold].index
    
    if len(rare_categories) > 0:
        df_new = df.copy()
        # Use a code that doesn't exist yet (max + 1)
        other_code = df[col].max() + 1
        df_new[col] = df_new[col].apply(lambda x: other_code if x in rare_categories else x)
        
        print(f"✓ {col}: Grouped {len(rare_categories)} rare categories → code {other_code}")
        return df_new
    else:
        print(f"✓ {col}: No rare categories found")
        return df

# Apply to all categorical features
for col in categorical_cols_updated:
    X_processed = group_rare_categories(X_processed, col, threshold=0.01, min_cardinality=10)
```

```python
# Check which features are highly imbalanced
print("\n" + "="*70)
print("Highly imbalanced features (>95% in one category):")
print("="*70)

imbalanced_features = []
for col in categorical_cols_updated:
    max_prop = X_processed[col].value_counts(normalize=True).max()
    if max_prop > 0.95:
        imbalanced_features.append(col)
        dominant_value = X_processed[col].value_counts().idxmax()
        print(f"⚠️ {col}: {max_prop*100:.1f}% are '{dominant_value}'")

print(f"\n💡 Recommendation: KEEP these {len(imbalanced_features)} imbalanced features")
print("   Reason: The minority class may be highly predictive!")
print("   Example: If 95% are NOT debtors, the 5% who ARE might have very different outcomes")
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize the scaler
scaler = StandardScaler()

# Scale numerical features
X_scaled = X.copy()
X_scaled[numerical_cols] = scaler.fit_transform(X[numerical_cols])

# Compare before and after
print("Before scaling:")
print(X[numerical_cols].head())
print("\nAfter scaling:")
print(X_scaled[numerical_cols].head())
```

```python
from sklearn.preprocessing import LabelEncoder

# Since all columns appear to be numerical already, 
# this is just for demonstration
if len(categorical_cols) > 0:
    le = LabelEncoder()
    for col in categorical_cols:
        X_scaled[col] = le.fit_transform(X[col])
    print(f"Encoded {len(categorical_cols)} categorical columns")
else:
    print("No categorical columns to encode - all features are already numerical!")
```

```python
from sklearn.model_selection import train_test_split

# Split the data (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, 
    test_size=0.2, 
    random_state=42, 
    stratify=y  # Ensures target distribution is preserved
)

print(f"Training set: {X_train.shape}")
print(f"Test set: {X_test.shape}")
print(f"\nTraining target distribution:\n{y_train.value_counts(normalize=True)}")
print(f"\nTest target distribution:\n{y_test.value_counts(normalize=True)}")
```

```python
# Visualize class distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Training set
axes[0].bar(y_train.value_counts().index, y_train.value_counts().values, 
            color=['red', 'orange', 'green'], alpha=0.7)
axes[0].set_title('Training Set - Target Distribution')
axes[0].set_xlabel('Target (0=Dropout, 1=Enrolled, 2=Graduate)')
axes[0].set_ylabel('Count')

# Test set
axes[1].bar(y_test.value_counts().index, y_test.value_counts().values, 
            color=['red', 'orange', 'green'], alpha=0.7)
axes[1].set_title('Test Set - Target Distribution')
axes[1].set_xlabel('Target (0=Dropout, 1=Enrolled, 2=Graduate)')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Check feature importance using correlation with target
feature_correlation = pd.DataFrame({
    'feature': numerical_cols,
    'correlation': [X_scaled[col].corr(y) for col in numerical_cols]
}).sort_values('correlation', key=abs, ascending=False)

print("Top 10 features by correlation with target:")
print(feature_correlation.head(10))

# Visualize
plt.figure(figsize=(10, 8))
plt.barh(feature_correlation['feature'][:15], feature_correlation['correlation'][:15])
plt.xlabel('Correlation with Target')
plt.title('Top 15 Features by Correlation')
plt.tight_layout()
plt.show()
```
